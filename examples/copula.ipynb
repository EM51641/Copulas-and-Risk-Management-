This notebook demonstrates how to use the copula library for risk analysis.
# 
The library provides tools for modeling dependencies between financial assets using copulas.
In this example, we'll:
1. Create a Gaussian copula model
2. Fit it to historical return data
3. Calculate Value at Risk (VaR) and Conditional Value at Risk (CVaR)
# 
The Gaussian copula is a popular choice for modeling financial dependencies
as it captures linear correlations between assets while being computationally
efficient.


```python
from yfinance import download
import numpy as np
import pandas as pd

# Load data
TICKERS =['CHFUSD=X', 'CHFEUR=X', 'CHFJPY=X', 'CHFCNY=X']
df = pd.DataFrame()
for ticker in TICKERS:
    df[ticker] = download(ticker, start="1998-01-01")['Close']
returns = df.pct_change().dropna()


# Convert returns to C-contiguous and correct dtype
# Returns are set to contiguous array to avoid memory issues as they are not in C-contiguous format
# when they come from pandas.
returns = np.ascontiguousarray(returns.astype(np.double))
WEIGHTS = np.array([0.25, 0.25, 0.25, 0.25], dtype=np.double)
```

```python
from copula.models.normal.gaussian import GaussianCopula

copula = GaussianCopula(
    WEIGHTS,
    returns=returns,
    size=1000000
)

copula.fit()

print("VaR: ", copula.var)
print("CVaR: ", copula.cvar)
```

# Student's t Copula for Risk Analysis
# 
The Student's t copula is another popular choice for modeling financial dependencies.
Unlike the Gaussian copula, it can capture both linear correlations and tail dependencies,
making it particularly suitable for modeling extreme market events.
# 
In this example, we'll:
1. Create a Student's t copula model
2. Fit it to historical return data
3. Calculate Value at Risk (VaR) and Conditional Value at Risk (CVaR)
# 
The Student's t copula is especially useful when dealing with financial returns
that exhibit heavy tails and tail dependence, which are common characteristics
of financial markets during periods of stress.


```python
from copula.models.student.student import StudentTCopula

copula = StudentTCopula(
    WEIGHTS,
    returns=returns,
    size=1000000
)

copula.fit()

print("VaR: ", copula.var)
print("CVaR: ", copula.cvar)
```

# Clayton Copula for Risk Analysis
# 
The Clayton copula is an asymmetric copula that exhibits strong lower tail dependence,
making it particularly suitable for modeling scenarios where assets tend to move
together during market downturns.
# 
In this example, we'll:
1. Create a Clayton copula model
2. Fit it to historical return data
3. Calculate Value at Risk (VaR) and Conditional Value at Risk (CVaR)
# 
The Clayton copula is especially useful when dealing with financial returns
that show asymmetric dependence, where the probability of extreme negative
movements occurring simultaneously is higher than extreme positive movements.


```python
from copula.models.clayton.clayton import ClaytonCopula

clayton_copula = ClaytonCopula(WEIGHTS, returns, size=100000)
clayton_copula.fit()

print("VaR: ", clayton_copula.var)
print("CVaR: ", clayton_copula.cvar)
```
